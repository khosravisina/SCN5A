```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import rdFingerprintGenerator
from rdkit import DataStructs
import numpy as np
import pandas as pd
from tqdm import tqdm
```

```python
list_columns = ['identifier', 'canonical_smiles', 'np_likeness', 'qed_drug_likeliness', 'annotation_level']

files = [
    "chebi-nps-04-2025.txt",
    "chembl-nps-04-2025.txt",
    "drugbanknp-04-2025.txt",
    "pubchem-nps-04-2025.txt",
    "zinc-np-04-2025.txt",
    ]

dataframes = []

for file in files:
    try:
        df = pd.read_table(file)
        # Select only the desired columns
        df_filtered = df[list_columns]
        dataframes.append(df_filtered)
    except Exception as e:
        print(f"Error reading {file}: {e}")

concatenated_df = pd.concat(dataframes, axis=0)

```

```python
concatenated_df['mol'] = concatenated_df['canonical_smiles'].apply(Chem.MolFromSmiles)
concatenated_df['canonical_smiles'] = concatenated_df['mol'].apply(Chem.MolToSmiles)
concatenated_df = concatenated_df.drop_duplicates('canonical_smiles')
concatenated_df[concatenated_df.duplicated('canonical_smiles')]
concatenated_df.to_csv("database-herbal-with-nplikeness.csv", index=False)
```
